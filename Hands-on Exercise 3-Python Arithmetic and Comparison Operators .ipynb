## Bonus Challenge 

```python
laptop_bag_price = 18500
wireless_mouse_price = 12000
keyboard_price = 24500
usb_drive_price = 5000
total_amount = laptop_bag_price + wireless_mouse_price + keyboard_price + usb_drive_price
print("Total Purchase:",total_amount)
discount_criteria = total_amount >= 50000
discount_amt = 0.1*total_amount
final_amount = total_amount - discount_amt
print("Discount Qualified:",discount_criteria)
print("Discount Amount:",discount_amt)
print("Final Amount:",final_amount)
```
