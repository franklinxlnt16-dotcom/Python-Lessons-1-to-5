## Exercise 2

```python
# creating variables and assigning values
employee_name = "David Musa"
department = "Finance"
basic_salary = 185000
salary_bonus = 35000
tax_rate = 0.075
gross_salary = basic_salary + salary_bonus
tax_amount = tax_rate * gross_salary
net_salary = gross_salary - tax_amount
# outputting the payroll system
print("="*36)
print("PAYROLL SUMMARY".center(36))
print("="*36)
print()
print(f'''Employee : {employee_name}

Department : {department}

Basic Salary : {basic_salary}

Bonus : {salary_bonus}

Gross Salary : {gross_salary}

Tax(7.5%) : {tax_amount}

Net Salary : {net_salary}''')
print()
print("="*36)
```

## Exercise 3

```python
customer_name = "Amina Bello"
product = "Wireless Headphones"
quantity = 3
unit_price = 27500
total_cost = quantity * unit_price
if quantity >2:
    discount = total_cost*0.1
else:
    discount = 0
final_cost = total_cost - discount
print("="*36)
print()
print("ONLINE STORE RECEIPT".center(36))
print()
print("="*36)
print()
print(f"""Customer : {customer_name}

Product : {product}

Quantity : {quantity }

Unit Price : {unit_price}

Total Cost : {total_cost}

Discount : {discount}

Final Amount : {final_cost}

Thank you for shopping with us!""")
print()
print("="*36)
```
